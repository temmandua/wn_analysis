```python
### first try on code to go through the hypernyms / hyponyms etc.
def synset_analysis(wordnet):
    synsets = wordnet.synsets()
    syndict = {}

    for s in synsets:
        hyponyms = s.hyponyms()
        hypolist = []
        for h in hyponyms:
            if h.lemmas() != []:
                hypolist.append(h.id)
                syndict[s] = hypolist

    return syndict

# bg.synset().lemmas()

bg_syndict = synset_analysis(bg)
```

```python
### first try on a code that gives you all the synsets that contain the lemma (regardless of sense)
### should we add pos?

def search_synsets(word, wordnet):
    synsets = wordnet.synsets()
    synset = {}

    for s in synsets:

        if word in s.lemmas():
            synset[s.id] = ''
            if s.hyponyms() != []:
                hyponyms = s.hyponyms()
                synset[s.id] = hyponyms



    return word, synset
```

```python
### new try... because the other code is very slow...
### we build an  index for the lemmas...

def build_lemma_index(wordnet):
    index = {}

    for s in wordnet.synsets():
        for lemma in s.lemmas():
            index.setdefault(lemma, []).append(s)
            # what exactly does setdefault do?
    return index
```

```python
### with this one, we go through the lemma index and

def search_synsets(word, index):
    synset = {}

    for s in index.get(word, ()):
        hyponyms = s.hyponyms()
        synset[s.id] = hyponyms if hyponyms else ''

    return word, synset
```

```python
### the mining code, with bg as example

lemma_index = build_lemma_index(bg)

results = {}

# also note that bg.lemmas() will be a list of words from openSubtitles
for lemma in bg.lemmas():
    results[lemma] = search_synsets(lemma, lemma_index)
```
