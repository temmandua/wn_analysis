```python
# console: pip install wn --upgrade
# --> wn
import wn
from wn import Wordnet
# terminal: python -m wn download omw:1.2
```

```python
### segment for defining all the Wordnets to work with

bg = wn.Wordnet('omw-bg:1.4')
pt = wn.Wordnet('omw-pt:1.4')
```

```python
### first try on code to go through the hypernyms / hyponyms etc.
def synset_analysis(wordnet):
    synsets = wordnet.synsets()
    syndict = {}

    for s in synsets:
        hyponyms = s.hyponyms()
        hypolist = []
        for h in hyponyms:
            if h.lemmas() != []:
                hypolist.append(h.lemmas())
        syndict[s] = hypolist

    return syndict

# bg.synset().lemmas()

bg_syndict = synset_analysis(bg)
```
